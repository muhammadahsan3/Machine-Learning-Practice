Missing Indicator

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split

from sklearn.impute import MissingIndicator,SimpleImputer
```

```python
df = pd.read_csv('train.csv',usecols=['Age','Fare','Survived'])
df.head()
```

```python
X = df.drop(columns=['Survived'])
y = df['Survived']
```

```python
X_train , X_test , y_train , y_test = train_test_split(X,y,test_size=0.2,random_state=2)
X_train.head()
```

```python
si = SimpleImputer()
X_train_trf = si.fit_transform(X_train)
X_test_trf = si.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

model = LogisticRegression()
model.fit(X_train_trf,y_train)
y_pred = model.predict(X_test_trf)
accuracy_score(y_test,y_pred)
```

Now using Missing indicator

```python
indicator = MissingIndicator()
indicator.fit(X_train)
```

```python
X_train_missing = indicator.transform(X_train)
X_test_missing = indicator.transform(X_test)
```

```python
X_train_missing
```

```python
X_train['Age_NA'] = X_train_missing
X_test['Age_NA'] = X_test_missing
```

```python
X_test.sample(5)
```

```python
si = SimpleImputer()
X_train_trf2 = si.fit_transform(X_train)
X_test_trf2 = si.transform(X_test)
```

```python
Model1 = LogisticRegression()
Model1.fit(X_train_trf2,y_train)
y_pred2 = Model1.predict(X_test_trf2)
accuracy_score(y_test,y_pred2)
```

we can also apply missing indicator with simple imputer

```python
si = SimpleImputer(add_indicator=True)
X_train = si.fit_transform(X_train)
X_test = si.transform(X_test)
```

```python
model3 = LogisticRegression()
model3.fit(X_train,y_train)
y_pred3 = model3.predict(X_test)
accuracy_score(y_test,y_pred3)
```
